After previewing some of the dataset these augmentations were selected:
*   hsv_h to adjust the colors to make the model robust to color variations
*   hsv_v to change brightness to simulate different lighting conditions
*   degrees to rotate images to handle different orientations.
*   shear to distort images to simulate perspective changes.

## Model evaluation

Evaluation Metrics Interpretation
Overall Model Performance
Precision (0.756): The model correctly predicted 75.6% of the detected objects.
Recall (0.762): The model successfully identified 76.2% of actual objects in the dataset.
mAP50 (0.763): At an IoU threshold of 50%, the model has 76.3% accuracy in localizing objects.
mAP50-95 (0.639): Across stricter IoU thresholds, the model's accuracy drops to 63.9%, indicating some difficulty in precise localization.
🔹 Augmentations did not significantly impact performance, suggesting that their values were not as influential as initially expected.


### Class-Specific Performance
Owl Detection: 🦉
* Precision (0.964): 96.4% of detected owls were correctly classified.
* Recall (0.967): The model correctly identified 96.7% of all owls.
* mAP50 (0.965): The model has 96.5% accuracy for localizing owls at a 50% overlap threshold.
* mAP50-95 (0.862): Even at stricter IoU thresholds, the model maintains 86.2% accuracy.
🔹 Conclusion: The model performs very well for detecting owls.

Sheep Detection: 🐑
* Precision (0.548): Only 54.8% of detected sheep were correctly classified.
* Recall (0.557): The model identified 55.7% of all sheep.
* mAP50 (0.560): Localization accuracy for sheep is 56.0% at a 50% overlap threshold.
*  mAP50-95 (0.416): Precision drops to 41.6% at stricter IoU thresholds.
🔹 Conclusion: The model struggles with detecting sheep, likely due to occlusion, background complexity, or appearance variability.


### Summary & Next Steps
*  The model performs very well for owls but struggles with sheep detection.
* Possible reasons for sheep detection issues:

Appearance variability (sheep in different poses, lighting, or backgrounds).
Occlusions making it harder for the model to detect sheep.
Less balanced dataset representation for sheep vs. owls.
Potential Improvements:
🔹 Experiment with different augmentation values to better adapt to sheep variations.
🔹 Optimize hyperparameters (e.g., learning rate, batch size).
🔹 Train for more epochs to allow better feature learning.
🔹 Class balancing: Ensure enough diverse samples for sheep in the dataset.
